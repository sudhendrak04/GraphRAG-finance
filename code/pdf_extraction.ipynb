```python
import fitz  # PyMuPDF
import pandas as pd
from pathlib import Path
```

```python
pdf_path = "D:/sudhendra/learning projects/GraphRAG/data/Apple23 report.pdf"
```

```python
doc = fitz.open(pdf_path)
```

```python
pages = []
for page_num,page in enumerate(doc,start=1):
    text = page.get_text()
    pages.append({
        "page" : page_num,
        "text" : text
    })

df = pd.DataFrame(pages)
df.head()
```
