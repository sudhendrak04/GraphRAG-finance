```python
import json
from pathlib import Path
import pandas as pd
```

```python
json_path = "D:/sudhendra/learning projects/GraphRAG/data/apple_2023_financial_metrics.json"

with open(json_path, "r") as f:
    metrics = json.load(f)

metrics.keys()
```

```python
def get_value(metric_key, year):
    return metrics[metric_key]["values"][str(year)]


def percentage_change(current, previous):
    return ((current - previous) / previous) * 100


def margin(part, total):
    return (part / total) * 100


def format_millions(value):
    return f"${value:,.0f} million"


def format_percent(value):
    return f"{value:.2f}%"
```

```python
revenue_2023 = get_value("total_net_sales", 2023)
revenue_2022 = get_value("total_net_sales", 2022)

revenue_change = revenue_2023 - revenue_2022
revenue_growth_pct = percentage_change(revenue_2023, revenue_2022)

print("Revenue 2023:", format_millions(revenue_2023))
print("Revenue 2022:", format_millions(revenue_2022))
print("Change:", format_millions(revenue_change))
print("Growth %:", format_percent(revenue_growth_pct))
```

```python
gross_margin_2023 = get_value("gross_margin", 2023)
operating_income_2023 = get_value("operating_income", 2023)
net_income_2023 = get_value("net_income", 2023)
rd_2023 = get_value("research_and_development", 2023)

gross_margin_pct = margin(gross_margin_2023, revenue_2023)
operating_margin_pct = margin(operating_income_2023, revenue_2023)
net_profit_margin_pct = margin(net_income_2023, revenue_2023)
rd_as_pct_sales = margin(rd_2023, revenue_2023)

print("Gross margin %:", format_percent(gross_margin_pct))
print("Operating margin %:", format_percent(operating_margin_pct))
print("Net profit margin %:", format_percent(net_profit_margin_pct))
print("R&D as % of sales:", format_percent(rd_as_pct_sales))
```

```python
sga_2023 = get_value("selling_general_and_administrative", 2023)
total_opex_2023 = get_value("total_operating_expenses", 2023)

sga_as_pct_sales = margin(sga_2023, revenue_2023)
opex_as_pct_sales = margin(total_opex_2023, revenue_2023)

print("SG&A as % of sales:", format_percent(sga_as_pct_sales))
print("Operating expenses as % of sales:", format_percent(opex_as_pct_sales))
```

```python
assets_2023 = get_value("total_assets", 2023)
liabilities_2023 = get_value("total_liabilities", 2023)
cash_2023 = get_value("cash_and_cash_equivalents", 2023)

liabilities_to_assets = margin(liabilities_2023, assets_2023)
cash_to_assets = margin(cash_2023, assets_2023)

print("Liabilities to assets:", format_percent(liabilities_to_assets))
print("Cash to assets:", format_percent(cash_to_assets))
```

```python
financial_calculations_2023 = {
    "company": "Apple",
    "year": 2023,
    "currency_unit": "USD millions",
    "revenue": revenue_2023,
    "revenue_change_from_2022": revenue_change,
    "revenue_growth_percent": round(revenue_growth_pct, 2),
    "gross_margin_percent": round(gross_margin_pct, 2),
    "operating_margin_percent": round(operating_margin_pct, 2),
    "net_profit_margin_percent": round(net_profit_margin_pct, 2),
    "r_and_d_as_percent_of_sales": round(rd_as_pct_sales, 2),
    "sga_as_percent_of_sales": round(sga_as_pct_sales, 2),
    "operating_expenses_as_percent_of_sales": round(opex_as_pct_sales, 2),
    "liabilities_to_assets_percent": round(liabilities_to_assets, 2),
    "cash_to_assets_percent": round(cash_to_assets, 2)
}

financial_calculations_2023
```

```python
output_path = "D:/sudhendra/learning projects/GraphRAG/data/apple_2023_financial_calculations.json"

with open(output_path, "w") as f:
    json.dump(financial_calculations_2023, f, indent=4)

output_path
```
